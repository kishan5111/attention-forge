# Grouped-Query Attention

Implement GQA and make the query-head to KV-head mapping explicit.

## Scope

- **Shape Contract:** GQA has `num_query_heads > num_kv_heads > 1`. Query heads are partitioned into groups.
- **Implementation Target:** Map each query head to a KV head, repeat K/V for attention, store only grouped K/V in cache.
- **Package Target:** Move the cleaned implementation into `attention_forge/attention/gqa.py`.

## Starter Cell

```python
def kv_head_for_query_head(query_head, num_query_heads, num_kv_heads):
    if num_query_heads % num_kv_heads != 0:
        raise ValueError("query heads must divide evenly by KV heads")
    group_size = num_query_heads // num_kv_heads
    return query_head // group_size


num_query_heads = 32
num_kv_heads = 8
mapping = {
    q: kv_head_for_query_head(q, num_query_heads, num_kv_heads)
    for q in range(num_query_heads)
}
cache_reduction_vs_mha = num_query_heads / num_kv_heads
mapping, cache_reduction_vs_mha
```
